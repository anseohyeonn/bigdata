## 문제 1. Tensor 5개 생성과 속성 확인표 (40점)
서로 다른 Tensor 5개를 만들고, 아래 항목을 담은 표를 출력하세요.

| 이름 | shape | dim | dtype | device |

각 Tensor를 왜 그렇게 만들었는지 주석으로 한 줄 적으세요.


```python
# 답안
import torch

# 1. 스칼라: 훈련 과정에서의 단일 손실값(loss)을 나타내기 위해 생성
t1 = torch.tensor(3.14)

# 2. 1차원 벡터: 3개의 연속형 변수/특성(feature)을 표현하기 위해 생성
t2 = torch.tensor([1.0, 2.5, 4.0])

# 3. 2차원 행렬: (배치 크기 2, 특성 수 3) 구조의 입력 데이터 배치를 생성하기 위해 zeros 사용
t3 = torch.zeros((2, 3), dtype=torch.int32)

# 4. 3차원 텐서: (배치 크기 2, 시퀀스 길이 4, 임베딩 차원 5) 구조의 시계열/텍스트 데이터를 난수로 표현하기 위해 randn 사용
t4 = torch.randn((2, 4, 5))

# 5. 4차원 텐서: (배치 크기 1, 채널 수 3, 높이 28, 너비 28) 구조의 이미지 데이터를 표현하기 위해 ones 사용
t5 = torch.ones((1, 3, 28, 28), dtype=torch.float64)

# 텐서 리스트 및 이름 설정
tensors = [("Tensor 1", t1), ("Tensor 2", t2), ("Tensor 3", t3), ("Tensor 4", t4), ("Tensor 5", t5)]

# 표 출력 (강의 자료에서 다루는 shape, dim, dtype, device 주요 속성 확인)
print(f"| {'이름':<8} | {'shape':<16} | {'dim':<4} | {'dtype':<15} | {'device':<8} |")
print("|" + "-"*10 + "|" + "-"*18 + "|" + "-"*6 + "|" + "-"*17 + "|" + "-"*10 + "|")

for name, t in tensors:
    # t.shape 대신 강의자료와 동일하게 t.dim() 및 주요 속성 확인
    print(f"| {name:<8} | {str(tuple(t.shape)):<16} | {t.dim():<4} | {str(t.dtype):<15} | {str(t.device):<8} |")
```

## 문제 2. Tensor 연산 (20점)
(3, 4) 와 (4, 2) Tensor를 만들어 아래를 출력하세요.
- 행렬곱 결과와 그 shape
- 첫 번째 Tensor의 열별 평균과 행별 합계


```python
# 답안
import torch

# (3, 4)와 (4, 2) 크기의 텐서 생성
a = torch.randn(3, 4)
b = torch.randn(4, 2)

# 1. 행렬곱 수행 (@ 연산자 또는 torch.matmul 사용)
# (3, 4) @ (4, 2) -> (3, 2)
matmul_result = a @ b

# 2. 첫 번째 Tensor(a)의 축(axis/dim) 연산
# 강의자료 2주차/4주차: axis(dim)=0 은 행을 따라 내려가며 계산하여 열별 통계가 남음 -> (4,)
col_mean = a.mean(dim=0)

# axis(dim)=1 은 열을 따라 옆으로 계산하여 행별 통계가 남음 -> (3,)
row_sum = a.sum(dim=1)

print("=== 1. 행렬곱 결과 및 shape ===")
print("행렬곱 결과:\n", matmul_result)
print("행렬곱 shape:", matmul_result.shape)

print("\n=== 2. 첫 번째 Tensor 연산 ===")
print("열별 평균 (dim=0):", col_mean, "| shape:", col_mean.shape)
print("행별 합계 (dim=1):", row_sum, "| shape:", row_sum.shape)
```

## 문제 3. CPU와 GPU 비교 (20점)
같은 크기의 행렬곱을 CPU와 GPU에서 각각 실행해 걸린 시간을 출력하고, 차이를 한두 줄로 설명하세요.
GPU를 쓸 수 없는 경우에는 CPU 결과와 그 이유를 적으세요.


```python
# 답안
import torch
import time

# 1. 실행 환경 확인
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"현재 실행 환경: {device}\n")

# 2. 연산용 행렬 생성 (5000 x 5000)
size = (5000, 5000)

# --- CPU 연산 측정 ---
x_cpu = torch.randn(size, device="cpu")
y_cpu = torch.randn(size, device="cpu")

start_time = time.time()
z_cpu = torch.matmul(x_cpu, y_cpu)
cpu_time = time.time() - start_time

print(f"CPU 실행 시간: {cpu_time:.4f}초")

# --- GPU 연산 측정 ---
x_gpu = x_cpu.to("cuda")
y_gpu = y_cpu.to("cuda")

# GPU 동기화 후 시간 측정
torch.cuda.synchronize()
start_time = time.time()
z_gpu = torch.matmul(x_gpu, y_gpu)
torch.cuda.synchronize()
gpu_time = time.time() - start_time

print(f"GPU 실행 시간: {gpu_time:.4f}초")
print(f"\n[속도 비교] GPU가 CPU보다 약 {cpu_time / gpu_time:.1f}배 빠릅니다.")
```

차이 설명: 일반적으로 GPU는 수많은 연산을 병렬로 처리하므로 5000×5000 행렬곱에서 CPU보다 빠른 경우가 많다. 실제 속도 차이는 하드웨어 성능과 메모리 전송 시간 등에 따라 달라진다.

## 문제 4. 자동미분 (20점)
아래 두 가지를 하세요.
1. $y = 2x^3 + 5x$ 에서 $x=3$ 일 때의 도함수를 자동미분으로 구하고, 손으로 구한 값과 비교하세요.
2. 자기가 만든 수식 하나에 대해 미분값을 구하고, 손 계산과 일치하는지 확인하세요.


```python
# 답안
import torch

print("=== [소문제 1] y = 2x^3 + 5x (x=3일 때) ===")

# 1. 미분 대상 변수 선언 (requires_grad=True가 핵심)
x1 = torch.tensor(3.0, requires_grad=True)

# 2. 수식 작성
y1 = 2 * (x1 ** 3) + 5 * x1

# 3. 역전파(자동미분) 실행
y1.backward()

# 4. 결과 출력 및 비교
print("자동미분 결과 (x1.grad) :", x1.grad.item())
print("손 계산 도함수 (6x^2 + 5): 6*(3^2) + 5 =", 6 * (3**2) + 5)
print("일치 여부               :", x1.grad.item() == (6 * (3**2) + 5))


print("\n=== [소문제 2] 임의의 수식: y = x^2 + 3x + 1 (x=4일 때) ===")

# 1. 미분 대상 변수 선언
x2 = torch.tensor(4.0, requires_grad=True)

# 2. 자기가 만든 수식 작성
y2 = (x2 ** 2) + 3 * x2 + 1

# 3. 역전파(자동미분) 실행
y2.backward()

# 4. 결과 출력 및 비교
print("자동미분 결과 (x2.grad) :", x2.grad.item())
print("손 계산 도함수 (2x + 3)  : 2*(4) + 3 =", 2 * 4 + 3)
print("일치 여부               :", x2.grad.item() == (2 * 4 + 3))
```
